## 1. Estructuras básicas de datos

En esta sección se utilizarán listas y diccionarios para representar
productos de Manager Stock. Estas estructuras permitirán organizar
información básica del inventario antes de convertirla en tablas
para su análisis.

```python
# Lista con los nombres de productos de prueba
productos = [
    "Arroz",
    "Leche",
    "Huevos",
    "Aceite",
    "Azúcar"
]

print("Productos registrados:", productos)
print("Cantidad de productos:", len(productos))
```

```python
# Registro individual de un producto de Manager Stock
producto = {
    "codigo": "PROD-001",
    "nombre": "Arroz",
    "categoria": "Granos",
    "precio": 4500,
    "stock": 25,
    "stock_minimo": 5
}

print("Registro del producto:")
print(producto)

print("Nombre:", producto["nombre"])
print("Precio:", producto["precio"])
print("Existencias:", producto["stock"])
```

```python
# Lista de diccionarios: cada diccionario representa un producto
datos_productos = [
    {
        "codigo": "PROD-001",
        "nombre": "Arroz",
        "categoria": "Granos",
        "precio": 4500,
        "stock": 25,
        "stock_minimo": 5
    },
    {
        "codigo": "PROD-002",
        "nombre": "Leche",
        "categoria": "Lácteos",
        "precio": 3800,
        "stock": 4,
        "stock_minimo": 10
    },
    {
        "codigo": "PROD-003",
        "nombre": "Huevos",
        "categoria": "Avícola",
        "precio": 16000,
        "stock": 15,
        "stock_minimo": 5
    },
    {
        "codigo": "PROD-004",
        "nombre": "Aceite",
        "categoria": "Aceites",
        "precio": 9500,
        "stock": 8,
        "stock_minimo": 5
    },
    {
        "codigo": "PROD-005",
        "nombre": "Azúcar",
        "categoria": "Granos",
        "precio": 4200,
        "stock": 3,
        "stock_minimo": 8
    }
]

for item in datos_productos:
    print(item["codigo"], "-", item["nombre"], "| Stock:", item["stock"])
```

### Análisis de las estructuras

La lista permite almacenar varios nombres de productos en un orden
determinado. El diccionario representa un producto mediante campos
como código, nombre, categoría, precio y existencias.

La lista de diccionarios permite organizar varios productos con la
misma estructura. En Manager Stock, esta representación puede servir
como punto de partida para preparar los datos del inventario y
analizar posteriormente los niveles de existencias.

## 2. NumPy aplicado al inventario

Se utilizará NumPy para analizar las cantidades disponibles de los
productos de prueba de Manager Stock mediante operaciones numéricas,
estadísticas básicas y filtros de existencias.

```python
# Extraer las existencias de los productos
existencias = np.array([
    producto["stock"] for producto in datos_productos
])

print("Existencias:", existencias)
print("Cantidad de productos:", len(existencias))
print("Existencias totales:", np.sum(existencias))
print("Promedio de existencias:", np.mean(existencias))
print("Mayor cantidad:", np.max(existencias))
print("Menor cantidad:", np.min(existencias))
```

```python
# Obtener el stock mínimo de cada producto
stocks_minimos = np.array([
    producto["stock_minimo"] for producto in datos_productos
])

# Comparar las existencias con el stock mínimo
productos_bajo_minimo = existencias < stocks_minimos

print("Stock mínimo:", stocks_minimos)
print("Productos por debajo del mínimo:", productos_bajo_minimo)

for i, producto in enumerate(datos_productos):
    if productos_bajo_minimo[i]:
        print(
            f"ALERTA: {producto['nombre']} - "
            f"Stock actual: {producto['stock']} - "
            f"Stock mínimo: {producto['stock_minimo']}"
        )
```

```python
# Simular una entrada de 5 unidades para cada producto
existencias_actualizadas = existencias + 5

print("Existencias originales:", existencias)
print("Existencias simuladas:", existencias_actualizadas)
```

### Conclusiones de NumPy

NumPy permitió calcular el total, el promedio, el máximo y el mínimo
de las existencias de los productos de prueba. También permitió
comparar las cantidades disponibles con los niveles mínimos
configurados para identificar productos que requieren revisión.

Las operaciones vectorizadas facilitan la ejecución de cálculos sobre
varios productos. En Manager Stock, este tipo de análisis puede apoyar
la detección temprana de necesidades de abastecimiento.

## 3. Introducción a Pandas: Series y DataFrames

Una Serie es una estructura unidimensional que permite almacenar
valores con un índice. Un DataFrame organiza los datos en filas y
columnas, facilitando el análisis tabular del inventario de Manager Stock.

```python
# Crear una Serie con los nombres de los productos
serie_productos = pd.Series([
    producto["nombre"] for producto in datos_productos
])

print("Serie de productos:")
display(serie_productos)
```

```python
# Convertir la lista de diccionarios en un DataFrame
df_productos = pd.DataFrame(datos_productos)

print("Tabla de productos de Manager Stock:")
display(df_productos)
```

```python
# Consultar columnas específicas
print("Nombres de los productos:")
display(df_productos["nombre"])

print("Productos con stock inferior al mínimo:")
display(
    df_productos[
        df_productos["stock"] < df_productos["stock_minimo"]
    ]
)
```

### Conclusiones de Pandas

La Serie permitió representar los nombres de los productos mediante
un índice. El DataFrame organizó los registros del inventario en filas
y columnas, facilitando la consulta de atributos específicos.

El filtrado de datos permitió identificar productos cuyas existencias
están por debajo del stock mínimo. Estas estructuras sirven como base
para generar reportes y apoyar las decisiones de abastecimiento en
Manager Stock.

### 3.7. Análisis del inventario de Manager Stock

En esta actividad se utilizarán filtros y operaciones de Pandas para identificar productos con existencias inferiores al stock mínimo, calcular el valor de cada producto almacenado y determinar el valor total del inventario. Este análisis permite reconocer necesidades de reposición y apoyar la toma de decisiones del negocio.

```python
# Identificar productos que necesitan reposición
productos_bajo_stock = df_productos[
    df_productos["stock"] < df_productos["stock_minimo"]
]

print("Productos que necesitan reposición:")
display(productos_bajo_stock)

# Calcular el valor del inventario por producto
df_productos["valor_inventario"] = (
    df_productos["precio"] * df_productos["stock"]
)

print("Valor del inventario por producto:")
display(df_productos)

# Calcular el valor total del inventario
valor_total = df_productos["valor_inventario"].sum()

print(f"Valor total del inventario: ${valor_total:,.0f} COP")

# Contar los productos que necesitan reposición
print(
    "Productos que necesitan reposición:",
    len(productos_bajo_stock)
)
```

### 3.8. Exportación de resultados procesados

Se exportarán los datos del inventario con el valor calculado por producto y un archivo independiente con los productos que requieren reposición. Estos archivos permitirán conservar los resultados del análisis y utilizarlos posteriormente en los reportes de Manager Stock.

```python
# Definir las rutas de salida
ruta_inventario_procesado = DATA_PROCESSED / "inventario_analizado.csv"
ruta_reposicion = DATA_PROCESSED / "productos_bajo_stock.csv"

# Exportar el inventario completo con el valor calculado
df_productos.to_csv(
    ruta_inventario_procesado,
    index=False,
    encoding="utf-8"
)

# Exportar los productos que necesitan reposición
productos_bajo_stock.to_csv(
    ruta_reposicion,
    index=False,
    encoding="utf-8"
)

# Verificar que los archivos se hayan creado
print("Inventario procesado:", ruta_inventario_procesado.exists())
print("Productos bajo stock:", ruta_reposicion.exists())

print("\nArchivos generados:")
print(ruta_inventario_procesado)
print(ruta_reposicion)
```
